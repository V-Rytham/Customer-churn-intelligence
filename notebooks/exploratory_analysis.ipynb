# Exploratory Analysis - Customer Churn
This notebook covers churn distribution, contract-level churn trends, tenure/churn patterns, and monthly charges analysis.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_processing import load_processed_data
from src.feature_engineering import add_engineered_features

df = add_engineered_features(load_processed_data())
df.head()
```

```python
sns.countplot(data=df, x='Churn')
plt.title('Churn Distribution')
plt.show()
```

```python
sns.countplot(data=df, x='Contract', hue='Churn')
plt.title('Churn by Contract Type')
plt.xticks(rotation=20)
plt.show()
```

```python
sns.boxplot(data=df, x='Churn', y='tenure')
plt.title('Tenure vs Churn')
plt.show()
```

```python
sns.boxplot(data=df, x='Churn', y='MonthlyCharges')
plt.title('Monthly Charges vs Churn')
plt.show()
```
